```python
import pandas as pd
import numpy as np

# Work on a copy; preserve the original dataframe
df_meteosat_work = df_meteosat.copy()

print("=" * 65)
print("METEOSAT — STAGE 1: MISSING-VALUE ASSESSMENT")
print("=" * 65)

# 1. Standard NaN / null values
missing_count = df_meteosat_work.isna().sum()

# 2. Empty or whitespace-only strings
empty_string_count = (
    df_meteosat_work.astype("string")
    .apply(lambda col: col.str.strip().eq("").sum())
)

# 3. Missing-value summary
missing_report = pd.DataFrame({
    "Missing_NaN": missing_count,
    "Empty_Strings": empty_string_count,
    "Total_Missing": missing_count + empty_string_count,
    "Missing_Percentage": (
        (missing_count + empty_string_count)
        / len(df_meteosat_work) * 100
    ).round(4)
})

display(missing_report)

# 4. Overall missingness
print("\nTotal NaN cells:", int(missing_count.sum()))
print("Total empty-string cells:", int(empty_string_count.sum()))
print("Rows containing at least one NaN:",
      int(df_meteosat_work.isna().any(axis=1).sum()))

# 5. Non-finite numeric values
numeric_df = df_meteosat_work.select_dtypes(include=np.number)

print("\nInfinite values by column:")
print(np.isinf(numeric_df).sum())

# 6. Verify original data remains unchanged
print("\nOriginal and working shapes:",
      df_meteosat.shape, df_meteosat_work.shape)
```

```python
import pandas as pd
import numpy as np

print("=" * 65)
print("METEOSAT — STAGE 2: DATA INTEGRITY")
print("=" * 65)

df_meteosat_integrity = df_meteosat_work.copy()

# 1. Exact duplicate observations
print("\n1. Duplicate observations")
print("Exact duplicate rows:",
      df_meteosat_integrity.duplicated().sum())

# 2. Construct timestamps without changing source columns
df_meteosat_integrity["Timestamp"] = pd.to_datetime(
    df_meteosat_integrity[
        ["Year", "Month", "Day", "Hour", "Minute"]
    ],
    errors="coerce"
)

print("\n2. Timestamp validity")
print("Invalid timestamps:",
      df_meteosat_integrity["Timestamp"].isna().sum())
print("First timestamp:",
      df_meteosat_integrity["Timestamp"].min())
print("Last timestamp:",
      df_meteosat_integrity["Timestamp"].max())

# 3. Chronological ordering
timestamps = df_meteosat_integrity["Timestamp"]

print("\n3. Chronological ordering")
print("Monotonically increasing:", timestamps.is_monotonic_increasing)
print("Duplicate timestamps:", timestamps.duplicated().sum())

# 4. Sampling interval
time_differences = timestamps.diff().dropna()

print("\n4. Sampling interval distribution")
print(time_differences.value_counts().head(10))

print("\nMost common interval:",
      time_differences.mode().iloc[0])

# 5. Unexpected time gaps
expected_interval = pd.Timedelta(minutes=15)

gap_report = time_differences[
    time_differences != expected_interval
]

print("\n5. Non-15-minute intervals")
print("Count:", len(gap_report))

if len(gap_report) > 0:
    print("\nFirst 20 deviations:")
    display(gap_report.head(20).to_frame("Time_Difference"))

# 6. Calendar coverage
print("\n6. Calendar coverage")
print("Unique dates:",
      timestamps.dt.date.nunique())
print("Unique months:",
      timestamps.dt.month.nunique())
print("Observations per month:")
print(timestamps.dt.month.value_counts().sort_index())

# 7. Original data preservation
print("\n7. Original data shape:", df_meteosat.shape)
print("Integrity copy shape:",
      df_meteosat_integrity.shape)
```

```python
import numpy as np
import pandas as pd

print("=" * 70)
print("METEOSAT — STAGE 3: SENSOR AND DATA-QUALITY ASSESSMENT")
print("=" * 70)

df_meteosat_quality = df_meteosat_integrity.copy()

sensor_columns = [
    "GHI", "DNI", "DHI", "Temperature",
    "Wind Speed", "Relative Humidity",
    "Pressure", "Solar Zenith Angle"
]

# 1. Descriptive statistics
print("\n1. Sensor descriptive statistics")
display(
    df_meteosat_quality[sensor_columns]
    .describe()
    .T
)

# 2. Negative and zero values
print("\n2. Negative and zero value counts")

value_checks = pd.DataFrame({
    "Negative_Count": (
        df_meteosat_quality[sensor_columns] < 0
    ).sum(),
    "Zero_Count": (
        df_meteosat_quality[sensor_columns] == 0
    ).sum()
})

display(value_checks)

# 3. IQR-based statistical outlier assessment
print("\n3. IQR-based outlier assessment")

outlier_records = []

for col in sensor_columns:
    q1 = df_meteosat_quality[col].quantile(0.25)
    q3 = df_meteosat_quality[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    count = (
        (df_meteosat_quality[col] < lower) |
        (df_meteosat_quality[col] > upper)
    ).sum()

    outlier_records.append({
        "Sensor": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "IQR_Outlier_Count": int(count),
        "Outlier_Percentage": round(
            count / len(df_meteosat_quality) * 100, 3
        )
    })

display(pd.DataFrame(outlier_records))

# 4. Solar irradiance and solar zenith behaviour
print("\n4. Solar behaviour checks")

daytime = df_meteosat_quality[
    df_meteosat_quality["Solar Zenith Angle"] < 90
]

nighttime = df_meteosat_quality[
    df_meteosat_quality["Solar Zenith Angle"] >= 90
]

print("Daytime observations:", len(daytime))
print("Nighttime observations:", len(nighttime))

print("\nMean irradiance — daytime")
display(daytime[["GHI", "DNI", "DHI"]].mean())

print("\nMean irradiance — nighttime")
display(nighttime[["GHI", "DNI", "DHI"]].mean())

# 5. Basic irradiance consistency check
print("\n5. GHI versus DNI + DHI exploratory comparison")

df_meteosat_quality["Estimated_GHI"] = (
    df_meteosat_quality["DNI"] *
    np.cos(np.radians(df_meteosat_quality["Solar Zenith Angle"]))
    + df_meteosat_quality["DHI"]
)

df_meteosat_quality["GHI_Residual"] = (
    df_meteosat_quality["GHI"] -
    df_meteosat_quality["Estimated_GHI"]
)

display(
    df_meteosat_quality[
        ["GHI", "Estimated_GHI", "GHI_Residual"]
    ].describe().T
)

print("\nQuality assessment completed. No observations were removed.")
```

```python
import pandas as pd
import numpy as np

print("=" * 70)
print("METEOSAT — STAGE 4: FEATURE AND TARGET SUITABILITY")
print("=" * 70)

df_meteosat_features = df_meteosat_quality.copy()

# 1. Create timestamp
df_meteosat_features["Timestamp"] = pd.to_datetime(
    df_meteosat_features[
        ["Year", "Month", "Day", "Hour", "Minute"]
    ]
)

# 2. Define feature roles
feature_roles = pd.DataFrame([
    ["GHI", "Solar irradiance", "Primary renewable-energy predictor / optional target"],
    ["DNI", "Solar irradiance", "Candidate predictor"],
    ["DHI", "Solar irradiance", "Candidate predictor"],
    ["Temperature", "Meteorological", "Candidate predictor"],
    ["Wind Speed", "Meteorological", "Candidate predictor"],
    ["Relative Humidity", "Meteorological", "Candidate predictor"],
    ["Pressure", "Meteorological", "Candidate predictor"],
    ["Solar Zenith Angle", "Solar geometry", "Candidate predictor"],
    ["Year", "Time metadata", "Timestamp construction only"],
    ["Month", "Time metadata", "Seasonality feature candidate"],
    ["Day", "Time metadata", "Timestamp construction only"],
    ["Hour", "Time metadata", "Time-of-day feature candidate"],
    ["Minute", "Time metadata", "Timestamp construction only"],
    ["Estimated_GHI", "Derived diagnostic", "Exclude initially; derived from DNI, DHI and zenith"],
    ["GHI_Residual", "Derived diagnostic", "Exclude initially; consistency diagnostic"],
], columns=[
    "Feature",
    "Category",
    "Proposed_Role"
])

display(feature_roles)

# 3. Correlation among measured environmental variables
print("\n1. Pearson correlation matrix")

sensor_columns = [
    "GHI", "DNI", "DHI", "Temperature",
    "Wind Speed", "Relative Humidity",
    "Pressure", "Solar Zenith Angle"
]

display(
    df_meteosat_features[sensor_columns].corr().round(3)
)

# 4. Seasonal distribution of key measurements
print("\n2. Monthly environmental and solar summary")

monthly_summary = (
    df_meteosat_features
    .groupby("Month")[
        ["GHI", "DNI", "DHI", "Temperature",
         "Wind Speed", "Relative Humidity"]
    ]
    .mean()
    .round(2)
)

display(monthly_summary)

# 5. Daytime versus nighttime feature distribution
print("\n3. Daytime and nighttime comparison")

df_meteosat_features["Solar_Period"] = np.where(
    df_meteosat_features["Solar Zenith Angle"] < 90,
    "Daytime",
    "Nighttime"
)

display(
    df_meteosat_features
    .groupby("Solar_Period")[
        ["GHI", "DNI", "DHI", "Temperature",
         "Wind Speed", "Relative Humidity"]
    ]
    .agg(["mean", "std"])
    .round(2)
)

# 6. Time-series suitability
print("\n4. Temporal coverage and candidate forecasting intervals")

print("Timestamp start:", df_meteosat_features["Timestamp"].min())
print("Timestamp end:", df_meteosat_features["Timestamp"].max())
print("Sampling interval: 15 minutes")
print("Potential forecasting horizons: 15, 30, 60 minutes")

# 7. Feature completeness
print("\n5. Candidate feature completeness")

display(
    df_meteosat_features[
        sensor_columns
    ].notna().mean().mul(100).round(2).to_frame(
        "Completeness_Percentage"
    )
)

print("\nNo features or observations have been removed.")
```

```python
import matplotlib.pyplot as plt
import numpy as np

correlation_columns = [
    "GHI",
    "DNI",
    "DHI",
    "Temperature",
    "Wind Speed",
    "Relative Humidity",
    "Pressure",
    "Solar Zenith Angle"
]

corr_matrix = df_meteosat_integrity[
    correlation_columns
].corr(method="pearson")

fig, ax = plt.subplots(figsize=(11, 8))

image = ax.imshow(
    corr_matrix,
    aspect="auto",
    vmin=-1,
    vmax=1,
    cmap="coolwarm"
)

ax.set_xticks(range(len(correlation_columns)))
ax.set_yticks(range(len(correlation_columns)))

ax.set_xticklabels(correlation_columns, rotation=45, ha="right")
ax.set_yticklabels(correlation_columns)

for i in range(len(correlation_columns)):
    for j in range(len(correlation_columns)):
        ax.text(
            j, i,
            f"{corr_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            fontsize=8
        )

plt.colorbar(image, ax=ax, label="Pearson correlation")
plt.title("METEOSAT — Sensor Correlation Heatmap")
plt.tight_layout()
plt.show()

print("Strong feature relationships (|r| >= 0.70):")

strong_pairs = []

for i in range(len(correlation_columns)):
    for j in range(i + 1, len(correlation_columns)):
        value = corr_matrix.iloc[i, j]

        if abs(value) >= 0.70:
            strong_pairs.append({
                "Feature_1": correlation_columns[i],
                "Feature_2": correlation_columns[j],
                "Correlation": round(value, 3)
            })

display(pd.DataFrame(strong_pairs))
```
